```python
import pandas as pd
from sklearn.dummy import DummyClassifier
from sklearn.pipeline import make_pipeline
from sklearn.compose import make_column_transformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

# Identificar colunas numéricas e categóricas
colunas_categoricas = ['Type']
colunas_numericas = ['Air temperature [K]', 'Process temperature [K]', 'Rotational speed [rpm]', 'Torque [Nm]', 'Tool wear [min]']

# Criar pré-processador compatível com os tipos de dados
preprocessador = make_column_transformer(
    (OneHotEncoder(drop='first'), colunas_categoricas),
    (StandardScaler(), colunas_numericas),
    remainder='passthrough'
)

# Definir os modelos utilizando pipelines para incluir o pré-processamento
modelos = {
    "Baseline": DummyClassifier(
        strategy="most_frequent"
    ),

    "Regressão Logística": make_pipeline(
        preprocessador,
        LogisticRegression(
            max_iter=2000,
            class_weight="balanced",
            random_state=42
        )
    ),

    "Árvore de Decisão": make_pipeline(
        preprocessador,
        DecisionTreeClassifier(
            max_depth=10,
            class_weight="balanced",
            random_state=42
        )
    ),

    "Random Forest": make_pipeline(
        preprocessador,
        RandomForestClassifier(
            n_estimators=100,
            class_weight="balanced",
            random_state=42,
            n_jobs=-1
        )
    )
}

# Treinar e avaliar os modelos
resultados = []
modelos_treinados = {}

for nome, modelo in modelos.items():
    # Treinamento
    modelo.fit(x_treino, y_treino)
    # Previsões
    previsoes = modelo.predict(x_teste)
    modelos_treinados[nome] = modelo

    resultados.append({
        "Modelo": nome,
        "Acurácia": accuracy_score(y_teste, previsoes),
        "Precisão": precision_score(y_teste, previsoes, zero_division=0),
        "Recall": recall_score(y_teste, previsoes, zero_division=0),
        "F1-score": f1_score(y_teste, previsoes, zero_division=0)
    })

# Criar e exibir tabela comparativa
tabela_resultados = pd.DataFrame(resultados)
display(
    tabela_resultados
    .sort_values("F1-score", ascending=False)
    .round(4)
)
```

```python
import pandas as pd
from sklearn.dummy import DummyClassifier
from sklearn.pipeline import make_pipeline
from sklearn.compose import make_column_transformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score

# 1. Definindo o pré-processamento correto para as features
colunas_categoricas = ['Type']
colunas_numericas = ['Air temperature [K]', 'Process temperature [K]', 'Rotational speed [rpm]', 'Torque [Nm]', 'Tool wear [min]']
preprocessador = make_column_transformer(
    (OneHotEncoder(drop='first'), colunas_categoricas),
    (StandardScaler(), colunas_numericas)
)

# 2. Definindo todos os 4 modelos para a tabela comparativa completa
modelos_completos = {
    "Baseline": DummyClassifier(strategy="most_frequent"),
    "Regressão Logística": make_pipeline(
        preprocessador,
        LogisticRegression(class_weight='balanced', max_iter=1000, random_state=42)
    ),
    "Árvore de Decisão": make_pipeline(
        preprocessador,
        DecisionTreeClassifier(max_depth=10, class_weight='balanced', random_state=42)
    ),
    "Random Forest": make_pipeline(
        preprocessador,
        RandomForestClassifier(n_estimators=100, class_weight='balanced', random_state=42, n_jobs=-1)
    )
}

resultados_completos = []

# 3. Treinando, prevendo e coletando métricas incluindo ROC-AUC
for nome, modelo in modelos_completos.items():
    modelo.fit(x_treino, y_treino)
    previsoes = modelo.predict(x_teste)

    # Calculando a ROC-AUC de maneira apropriada para cada tipo de classificador
    if hasattr(modelo, "predict_proba"):
        probabilidades = modelo.predict_proba(x_teste)[:, 1]
    else:
        # Para modelos que não estimam probabilidades (como algumas estratégias do Dummy)
        probabilidades = previsoes

    auc_score = roc_auc_score(y_teste, probabilidades)

    resultados_completos.append({
        'Modelo': nome,
        'Acurácia': accuracy_score(y_teste, previsoes),
        'Precisão': precision_score(y_teste, previsoes, zero_division=0),
        'Recall': recall_score(y_teste, previsoes, zero_division=0),
        'F1-score': f1_score(y_teste, previsoes, zero_division=0),
        'ROC-AUC': auc_score
    })

# 4. Criando e exibindo a tabela unificada final ordenada pelo F1-score
tabela_completa_auc = pd.DataFrame(resultados_completos)
display(tabela_completa_auc.sort_values(by='F1-score', ascending=False).round(4))
```

```python
import numpy as np
import pandas as pd

from sklearn.model_selection import StratifiedKFold, cross_val_predict
from sklearn.metrics import (
    precision_score,
    recall_score,
    f1_score
)

# Validação cruzada nos dados de treino
cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

# Probabilidades previstas para cada observação
# usando validação cruzada
probabilidades = cross_val_predict(
    melhor_rf,
    x_treino,
    y_treino,
    cv=cv,
    method="predict_proba",
    n_jobs=-1
)

# Localizar a coluna correspondente à classe "Com falha" (1)
indice_falha = list(melhor_rf.classes_).index(1)
prob_falha = probabilidades[:, indice_falha]

# Comparar vários limiares
resultados = []

for limiar in np.arange(0.10, 0.66, 0.05):
    previsoes = (prob_falha >= limiar).astype(int)

    resultados.append({
        "Limiar": round(limiar, 2),
        "Precisao": precision_score(
            y_treino, previsoes, zero_division=0
        ),
        "Recall": recall_score(
            y_treino, previsoes, zero_division=0
        ),
        "F1-score": f1_score(
            y_treino, previsoes, zero_division=0
        )
    })

tabela_limiares = pd.DataFrame(resultados)

display(
    tabela_limiares.round(3).sort_values(
        "Recall", ascending=False
    )
)
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.compose import make_column_transformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import roc_auc_score

# 1. Carregar o dataset caso as variáveis não estejam prontas
url = 'https://archive.ics.uci.edu/ml/machine-learning-databases/00601/ai4i2020.csv'
df = pd.read_csv(url)

# 2. Separar recursos e alvo
alvo = 'Machine failure'
colunas_excluir = ['UDI', 'Product ID', 'Machine failure', 'TWF', 'HDF', 'PWF', 'OSF', 'RNF']
x = df.drop(columns=[c for c in colunas_excluir if c in df.columns])
y = df[alvo]

# 3. Divisão de treino e teste
x_treino, x_teste, y_treino, y_teste = train_test_split(
    x, y, test_size=0.20, random_state=42, stratify=y
)

# 4. Pipeline de Preprocessamento
colunas_categoricas = ['Type']
colunas_numericas = ['Air temperature [K]', 'Process temperature [K]', 'Rotational speed [rpm]', 'Torque [Nm]', 'Tool wear [min]']
preprocessador = make_column_transformer(
    (OneHotEncoder(drop='first'), colunas_categoricas),
    (StandardScaler(), colunas_numericas)
)

# 5. Inicializar e treinar o Random Forest
melhor_rf = make_pipeline(
    preprocessador,
    RandomForestClassifier(n_estimators=200, class_weight='balanced', random_state=42, n_jobs=-1)
)
melhor_rf.fit(x_treino, y_treino)

# 6. Calcular a probabilidade da classe de falha (classe 1)
y_prob = melhor_rf.predict_proba(x_teste)[:, 1]

# 7. Cálculo da ROC-AUC
roc_auc = roc_auc_score(y_teste, y_prob)
print(f"ROC-AUC do modelo Random Forest: {roc_auc:.4f}")
```

```python
alvo = 'Machine failure'
colunas_excluir = ['UDI', 'Product ID', 'Machine failure', 'TWF', 'HDF', 'PWF', 'OSF', 'RNF']

# Filtrando apenas colunas que realmente existem no DataFrame
x = df.drop(columns=[c for c in colunas_excluir if c in df.columns])
y = df[alvo]

print('Features (X) selecionadas:')
print(x.columns.tolist())
print('\nFormato de X:', x.shape)
print('Formato de y:', y.shape)
```

```python
from sklearn.model_selection import train_test_split

# Realizando a divisão em treino e teste (80% treino, 20% teste) de forma estratificada
x_treino, x_teste, y_treino, y_teste = train_test_split(
    x, y, test_size=0.20, random_state=42, stratify=y
)

print('Registros para treinamento:', len(x_treino))
print('Registros para teste:', len(x_teste))
```

```python
from sklearn.pipeline import make_pipeline
from sklearn.compose import make_column_transformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier

# Identificando colunas categóricas e numéricas para o pré-processamento
colunas_categoricas = ['Type']
colunas_numericas = ['Air temperature [K]', 'Process temperature [K]', 'Rotational speed [rpm]', 'Torque [Nm]', 'Tool wear [min]']

# Criando o pré-processador
preprocessador = make_column_transformer(
    (OneHotEncoder(drop='first'), colunas_categoricas),
    (StandardScaler(), colunas_numericas)
)

# Definindo o dicionário com os modelos corrigidos dentro de pipelines
modelos = {
    'Regressão Logística': make_pipeline(
        preprocessador,
        LogisticRegression(class_weight='balanced', max_iter=1000, random_state=42)
    ),
    'Random Forest': make_pipeline(
        preprocessador,
        RandomForestClassifier(n_estimators=200, class_weight='balanced', random_state=42, n_jobs=-1)
    )
}

# Treinando cada modelo
for nome, modelo in modelos.items():
    modelo.fit(x_treino, y_treino)
    print(f'{nome} - treinamento concluído.')
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
import pandas as pd

resultados = []

# Gerando previsões e calculando as métricas para cada modelo treinado
for nome, modelo in modelos.items():
    previsoes = modelo.predict(x_teste)

    resultados.append({
        'Modelo': nome,
        'Acurácia': accuracy_score(y_teste, previsoes),
        'Precisão': precision_score(y_teste, previsoes, zero_division=0),
        'Recall': recall_score(y_teste, previsoes, zero_division=0),
        'F1-score': f1_score(y_teste, previsoes, zero_division=0)
    })

# Criando o DataFrame com os resultados obtidos
tabela_resultados = pd.DataFrame(resultados)

# Exibindo os resultados ordenados pelo F1-score de forma decrescente
display(tabela_resultados.sort_values(by='F1-score', ascending=False).round(4))
```

```python
from sklearn.metrics import classification_report, confusion_matrix

# Identificando o nome do melhor modelo dinamicamente pelo maior F1-score
melhor_nome = tabela_resultados.sort_values(by='F1-score', ascending=False).iloc[0]['Modelo']
melhor_modelo = modelos[melhor_nome]

# Gerando as previsões com o melhor modelo
previsoes = melhor_modelo.predict(x_teste)

print(f"Melhor modelo pelo F1-score: {melhor_nome}\n")

print("Relatório de Classificação:")
print(classification_report(y_teste, previsoes, zero_division=0))

print("Matriz de Confusão:")
print(confusion_matrix(y_teste, previsoes))
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

nome_rf = 'Random Forest'
if nome_rf in modelos:
    modelo_rf_pipeline = modelos[nome_rf]

    # Extraindo o classificador final do pipeline
    classificador_rf = modelo_rf_pipeline.named_steps['randomforestclassifier']

    # Extraindo o pre-processador para recuperar os novos nomes das colunas após o OneHotEncoder
    trans = modelo_rf_pipeline.named_steps['columntransformer']

    # Obtendo os nomes das colunas geradas pelo OneHotEncoder e mantendo as numéricas
    nomes_categoricos = list(trans.named_transformers_['onehotencoder'].get_feature_names_out(colunas_categoricas))
    nomes_colunas_finais = nomes_categoricos + colunas_numericas

    # Mapeando as importâncias das features
    importancias = pd.Series(classificador_rf.feature_importances_, index=nomes_colunas_finais).sort_values(ascending=False)

    print("Variáveis mais importantes:")
    display(importancias.to_frame('Importância'))

    # Plotando o gráfico de barras horizontais das importâncias
    plt.figure(figsize=(10, 6))
    importancias.sort_values().plot(kind='barh', color='skyblue')
    plt.title('Importância das Variáveis - Random Forest')
    plt.xlabel('Importância')
    plt.ylabel('Variáveis')
    plt.tight_layout()
    plt.show()
else:
    print("O modelo Random Forest não foi encontrado no dicionário.")
```

```python
# Extraindo os valores diretamente da matriz de confusão (cm)
# cm[0, 0] = VN, cm[0, 1] = FP, cm[1, 0] = FN, cm[1, 1] = VP
vn, fp, fn, vp = cm[0, 0], cm[0, 1], cm[1, 0], cm[1, 1]

# Cálculo das métricas
precisao = vp / (vp + fp) if (vp + fp) > 0 else 0
recall = vp / (vp + fn) if (vp + fn) > 0 else 0
especificidade = vn / (vn + fp) if (vn + fp) > 0 else 0
f1 = 2 * (precisao * recall) / (precisao + recall) if (precisao + recall) > 0 else 0
acuracia = (vp + vn) / (vp + vn + fp + fn)

# Exibição dos resultados formatados
print("--- Métricas Calculadas Manualmente ---")
print(f"Acurácia: {acuracia:.2%}")
print(f"Precisão: {precisao:.2%}")
print(f"Recall (Sensibilidade): {recall:.2%}")
print(f"Especificidade: {especificidade:.2%}")
print(f"F1-score: {f1:.2%}")
```
